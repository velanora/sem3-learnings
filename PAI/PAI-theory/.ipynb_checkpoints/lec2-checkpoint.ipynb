# 4. Development Environment Setup
<br>

A development environment is a collection of software tools (Code Editor, Interpreter, Debugger, Terminal, Package Manager etc.) used to write, run, test, and debug programs.


## Explanation

- Python must be installed before writing programs.
- An editor or IDE helps create and manage code efficiently.
- Package managers (pip etc.) install additional libraries required for AI projects.
- Jupyter Notebook provides an interactive environment for AI and Data Science.
- Proper setup improves productivity and reduces configuration issues.
- Virtual environments help isolate project dependencies. (detail in next slide)

### Virtual Environments

**Virtual environments** are isolated Python environments created for individual projects. They allow each project to have its own set of Python packages and package versions without affecting other projects or the system-wide Python installation.

#### What does "isolate project dependencies" mean?

Suppose you have two Python projects:

- **Project A** requires `pandas 2.2`
- **Project B** requires `pandas 3.0`

If you install both packages globally, you may run into **version conflicts** because one project might not work correctly with the version required by the other.


With virtual environments:

```text
System Python
│
├── Project A
│   └── Virtual Environment
│       ├── pandas 2.2
│       └── numpy ...
│
└── Project B
    └── Virtual Environment
        ├── pandas 3.0
        └── numpy ...
```

## Basic Setup Steps

Run the following commands on terminal(Linux) after installing Python.

### 1. Verify Python

```bash
python3 --version
```

### 2. Install Jupyter Notebook

```bash
pip install notebook
```

### 3. Start Jupyter Notebook

```bash
jupyter notebook
```

### 4. Install NumPy

```bash
pip install numpy
```
