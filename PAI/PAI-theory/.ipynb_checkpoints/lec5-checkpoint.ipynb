## 1. Logical Operators

<br>

Logical operators are used to combine multiple conditions and return a Boolean value (`True` or `False`).



### Explanation

- Used in decision-making statements (`if`, `while`).
- Combine two or more conditions.
- Return `True` or `False`.
- Commonly used in AI, data filtering, and validation.
- Python provides three logical operators: `and`, `or`, and `not`.

### Operators

| Operator | Meaning |
|----------|---------|
| `and` | True if both conditions are true |
| `or` | True if at least one condition is true |
| `not` | Reverses the result |


### Example


````python

x = 10
y = 20

print(x > 5 and y > 15)   
print(x > 15 or y > 15) 
print(not(x > 5))        
````

# 2. Assignment Operators

<br>

Assignment operators are used to assign values to variables and update their values.

## Explanation

- Simplify arithmetic operations with assignment.
- Reduce code length.
- Improve readability.
- Commonly used in loops and counters.
- The variable on the left receives the result.

## Common Assignment Operators

| Operator | Example | Equivalent To | Short Description |
|----------|---------|----------------|-------------------|
| `=` | `x = 5` | Assign 5 to `x` | Assigns a value to a variable |
| `+=` | `x += 3` | `x = x + 3` | Adds a value and assigns the result |
| `-=` | `x -= 3` | `x = x - 3` | Subtracts a value and assigns the result |
| `*=` | `x *= 3` | `x = x * 3` | Multiplies by a value and assigns the result |
| `/=` | `x /= 3` | `x = x / 3` | Divides by a value and assigns the result |
| `%=` | `x %= 3` | `x = x % 3` | Assigns the remainder of division |
| `**=` | `x **= 2` | `x = x ** 2` | Raises to a power and assigns the result |
| `//=` | `x //= 2` | `x = x // 2` | Performs floor division and assigns the result |

## Example


````python
x = 10

x += 5
print(x)

x *= 2
print(x)

x -= 4
print(x)
````

````python
x = 10

x **= 2
print(x)

y = x

x /= 3
print(x)

y //= 3
print(y)
````

# 3. Identity Operators

<br>

Identity operators check whether two variables refer to the **same object in memory**, not just whether their values are equal.


## Explanation

- Compare object identity.
- Useful when working with lists, dictionaries, and objects.
- Different from `==`, which compares values.
- Frequently used to check for `None`.
None is a special value in Python that means "no value" or "nothing".
- Python provides `is` and `is not` for checking object identity.

````python
name = None
# Here, name does not contain a person's name. It currently has no value.

if name is None:
    print("No name was provided")
````

## Operators

| Operator | Meaning |
|----------|---------|
| `is` | True if both refer to the same object |
| `is not` | True if they refer to different objects |


## Example 1: Same Object

````python
a = [1, 2, 3]
b = a

print(a is b)
print(a == b)
print(a != b)
````

## Example 2: Different Objects with Same Values


````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
````

## Example 3: Checking for `None`

````python
value = None

print(value is None)
````

````python
value = 2

print(value is 2)
````

# 4. Membership Operators

<br>

Membership operators check **whether a value exists within a sequence** such as a string, list, tuple, set, or dictionary.


## Explanation

- Used to search for an element.
- Return `True` or `False`.
- Commonly used in validation and filtering.
- Work with strings, lists, tuples, sets, and dictionaries.
- Python provides `in` and `not in`.


## Operators

| Operator | Meaning |
|----------|---------|
| `in` | True if the value exists |
| `not in` | True if the value does not exist |

## Example 1: List Membership

````python
numbers = [10, 20, 30, 40]

print(20 in numbers)
print(20 not in numbers)
print(50 in numbers)
````

## Example 2: String Membership

````python
text = "Artificial Intelligence"

print("AI" in text)
print("Intel" in text)
````

> **Note:** Membership in strings is **case-sensitive** and checks for an exact substring.  
> `"AI"` is **not** present in `"Artificial Intelligence"`.

## Example 3: Dictionary Membership

````python
student = {
    "name": "Ali",
    "age": 20
}

print(20 in student)
print("marks" not in student)
````

# Quick Comparison

| Category | Operators | Purpose |
|-----------|-----------|---------|
| Logical | `and`, `or`, `not` | Combine conditions |
| Assignment | `=`, `+=`, `-=`, `*=`, `/=`, `%=`, `**=`, `//=` | Assign and update values |
| Identity | `is`, `is not` | Check if two variables refer to the same object |
| Membership | `in`, `not in` | Check whether an item exists in a collection |


# C/C++ vs Python Comparison

| Concept | C/C++ | Python |
|---------|-------|--------|
| Logical AND | `&&` | `and` |
| Logical OR | `\|\|` | `or` |
| Logical NOT | `!` | `not` |
| Assignment Add | `+=` | `+=` |
| Identity Check | No direct equivalent | `is`, `is not` |
| Membership Check | Manual loop/search | `in`, `not in` |

# Key Takeaways

- **Logical operators** combine multiple conditions.
- **Assignment operators** update variable values efficiently.
- **Identity operators (`is`, `is not`)** compare object identities (memory locations), not values.
- **Membership operators (`in`, `not in`)** make searching within collections simple and readable.
- Python's `is` and `in` operators have **no direct equivalents in C/C++**, making them unique and widely used features of Python.
