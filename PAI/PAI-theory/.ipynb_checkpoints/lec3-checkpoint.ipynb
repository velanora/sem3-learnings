# 1. From C/C++ to Python

<br>

From **C/C++ to Python** refers to understanding how Python differs from traditional compiled languages such as C and C++, and learning the new programming style used in Python.



### Key Points

- Python is a **high-level, interpreted language**, whereas C/C++ are generally **compiled languages**.
- Python focuses on **readability and simplicity**, reducing the amount of code required.
- Variables **do not require explicit type declarations**.
- Python uses **indentation** instead of curly braces `{}` to define code blocks.
- Python is widely used in:
  - Artificial Intelligence (AI)
  - Machine Learning (ML)
  - Data Science
  - Automation
  - Web Development


## Comparison Example

### C++

```cpp
#include <iostream>
using namespace std;

int main()
{
    int age = 20;
    cout << age << endl;
    return 0;
}
```

### Python

```python
age = 20
print(age)
```

## Major Differences

| Feature | C/C++ | Python |
|----------|--------|---------|
| Compilation | Compiled | Interpreted |
| Variable Declaration | Required | Not Required |
| Semicolon (`;`) | Required | Not Required |
| Curly Braces (`{}`) | Used | Not Used |
| Indentation | Optional | Mandatory |
| Memory Management | Manual (`new/delete`) | Automatic |
| Syntax | More Verbose | More Concise |

## Variables, Naming Rules, Basic Data Types, Input and Output

# 4. Variables in Python

## Definition

A **variable** is a named memory location used to store data that can be accessed and modified during program execution.

## Explanation

- Variables store values such as numbers, text, or logical values.
- Python creates a variable automatically when a value is assigned.
- No explicit type declaration is required.
- A variable's value can be changed at any time.
- Variables make programs flexible and reusable.

## Example

````python
x = 10
name = "Ali"
price = 99.5

print(x)
print(name)
print(price)
````

## Dynamic Typing Example

````python
x = 10
print(x)

x = "Hello"
print(x)
````

# 5. Naming Rules for Variables
<br>

**Naming rules** are the guidelines that determine how identifiers (variable names, function names, etc.) can be created in Python.



## Explanation

- Variable names can contain letters (`A-Z`, `a-z`), digits (`0-9`), and underscores (`_`).
- A variable name **cannot start with a digit**.
- Variable names are **case-sensitive**.
- Reserved keywords cannot be used as variable names.
- Meaningful names improve code readability and maintenance.
- Python follows the **snake_case** naming convention.


## Valid Variable Names

````python
student_name = "Ahmed"
age = 20
_marks = 90
total_amount = 500
````

## Invalid Variable Names

````python
2age = 20          # Starts with digit
student-name = ""  # Hyphen not allowed
class = 5          # Keyword
````

## Case Sensitivity Example


````python
name = "Ali"
Name = "Ahmed"

print(name)
print(Name)
````

## Good Naming Practice


````python
total_salary = 50000
student_count = 120
average_marks = 85
````

## Poor Naming Practice

````python
x = 50000
a = 120
b = 85
````

# 6. Basic Data Types in Python

## Definition

A **data type** specifies the kind of value a variable can store and the operations that can be performed on it.



## Explanation

- Python provides several built-in data types.
- Different data types occupy memory differently.
- Operations depend on the type of data.
- Python automatically determines the data type.
- The `type()` function can be used to check the type of a variable.


## Integer (`int`)

Stores whole numbers.

````python
age = 25
print(type(age))
````

## Float (`float`)

Stores decimal numbers.

````python
pi = 3.14159
print(type(pi))
````

## String (`str`)

Stores text enclosed in quotes.


````python
name = "Python"
print(type(name))
````

## Boolean (`bool`)

Stores logical values (`True` or `False`).

````python
is_passed = True
print(type(is_passed))
````

## Multiple Data Types Example


````python
name = "Ali"
age = 21
cgpa = 3.7
is_graduated = False

print(type(name))
print(type(age))
print(type(cgpa))
print(type(is_graduated))
````

## Summary Table

| Data Type | Description | Example |
|-----------|-------------|---------|
| `int` | Whole numbers | `10` |
| `float` | Decimal numbers | `3.14` |
| `str` | Text | `"Hello"` |
| `bool` | Logical values | `True` |

# 7. Input and Output in Python

## Definition

**Input** is the data provided to a program, while **output** is the information produced by the program.

## Explanation

- Python uses the `input()` function to receive data from the user.
- Python uses the `print()` function to display output.
- Data entered through `input()` is received as a **string** by default.
- Type conversion may be required for numerical calculations.
- Input and output enable interaction between users and programs.


## Output Using `print()`

````python
print("Welcome to Python")
````

## Taking Input

````python
name = input("Enter your name: ")

print("Hello", name)
````

## Numeric Input

````python
age = int(input("Enter your age: "))

print("Age =", age)
````

## Taking Two Numbers

````python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

sum = a + b

print("Sum =", sum)
````

## Formatted Output (f-Strings)

````python
name = "Ali"
age = 20

print(f"My name is {name} and I am {age} years old.")
````

# Quick Comparison with C++

| Feature | C++ | Python |
|----------|-----|---------|
| Variable Declaration | `int x = 10;` | `x = 10` |
| Input | `cin >> x;` | `x = input()` |
| Numeric Input | `cin >> x;` | `x = int(input())` |
| Output | `cout << x;` | `print(x)` |
| Type Specification | Required | Automatic |
| String | `string name;` | `name = "Ali"` |

# Key Takeaways

- Variables store data and are created automatically upon assignment.
- Python uses **dynamic typing**, allowing a variable to change its data type during execution.
- Variable names must follow Python's naming rules.
- The four fundamental data types are `int`, `float`, `str`, and `bool`.
- `input()` is used to receive data from the user.
- `print()` is used to display output.
- `input()` always returns a **string**, so functions like `int()` and `float()` are needed for numerical input.
