## 1. `if` Statement 
<br>
The `if` statement is a decision-making construct that executes a block of code only when a specified condition evaluates to `True`.

### Explanation

- Used to make decisions in a program.
- The condition must evaluate to either `True` or `False`.
- If the condition is `True`, the indented block executes.
- If the condition is `False`, the block is skipped.
- Multiple independent `if` statements can exist in a program.


### Syntax

```python
if condition:
    statements
```


### Example

````python
age = 20

if age >= 18:
    print("You are eligible to vote.")
````

### Comparison with C/C++

#### C/C++

```cpp
if(age >= 18)
{
    cout << "Eligible";
}
```


#### Python

```python
if age >= 18:
    print("Eligible")
```


## 2. `if-else` Statement

### Definition

The `if-else` statement provides an alternative path of execution when the condition is `False`.


### Explanation

- Executes one block if the condition is `True`.
- Executes another block if the condition is `False`.
- Useful when exactly two possible outcomes exist.
- Improves readability compared to multiple separate `if` statements.
- Commonly used for validations and comparisons.

### Syntax

```python
if condition:
    statements
else:
    statements
```


### Example

````python
number = 7

if number % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
````

### Flow

```text
Condition True  -> if block
Condition False -> else block
```

## 3. `if-elif-else` Statement

### Definition

The `if-elif-else` statement is used when multiple conditions need to be tested sequentially.



### Explanation

- `elif` means "else if".
- Conditions are checked from top to bottom.
- The first true condition executes.
- Remaining conditions are skipped.
- Optional `else` block handles all unmatched cases.


### Syntax

```python
if condition1:
    statements
elif condition2:
    statements
elif condition3:
    statements
else:
    statements
```

### Example: Marks Check

````python
marks = 78

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Grade D")
````

### Example: Temperature Check

````python
temperature = 35

if temperature > 40:
    print("Very Hot")
elif temperature > 30:
    print("Hot")
elif temperature > 20:
    print("Pleasant")
else:
    print("Cold")
````

## 4. Nested `if` Statements

### Definition

A nested `if` statement is an `if` statement placed inside another `if`, `elif`, or `else` block.



### Explanation

- Used for hierarchical decision making.
- Allows checking secondary conditions.
- Common in authentication and validation systems.
- Can become difficult to read if nested too deeply.
- Usually replaced by logical operators when possible.


### Example


````python

age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to vote")
````

## 5. Logical Operators with Conditions
<br>
Logical operators combine multiple conditions into a single expression.



### Common Operators

| Operator | Meaning |
|----------|---------|
| `and` | Both conditions must be True |
| `or` | At least one condition must be True |
| `not` | Reverses the condition |


### Example

````python
age = 22
has_license = True

if age >= 18 and has_license:
    print("Can drive")
````

### Example

````python
temperature = 15

if not temperature > 20:
    print("Temperature is 20 or below")
````

## 6. Ternary Conditional Operator

Python provides a shorthand form of `if-else` called the conditional expression or ternary operator.


### Explanation


- Useful for simple decisions.
- Written in a single line.
- Improves code compactness.
- Should be used only for simple expressions.


### Syntax

```python
value_if_true if condition else value_if_false
```


### Example

````python
age = 17

status = "Adult" if age >= 18 else "Minor"
print(status)
````

## 7. Match-Case Statement (Python 3.10+)


The `match-case` statement is Python's pattern-matching construct, similar in purpose to the `switch-case` statement found in languages such as C, C++, and Java.



### Explanation

- Introduced in Python 3.10.
- Matches a value against multiple patterns.
- Executes the first matching case.
- Improves readability when many choices exist.
- `_` acts as the default case.


### Syntax

```python
match variable:
    case value1:
        statements
    case value2:
        statements
    case _:
        statements
```


### Example

````python
day = 3

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid Day")
````

### Comparison with C/C++ Switch

#### C/C++

```cpp
switch(day)
{
    case 1:
        cout << "Monday";
        break;
    case 2:
        cout << "Tuesday";
        break;
    default:
        cout << "Invalid";
}
```

#### Python

```python
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case _:
        print("Invalid")
```


## 8. Match-Case with Multiple Values

### Example


````python
day = "Saturday"

match day:
    case "Saturday" | "Sunday":
        print("Weekend")
    case _:
        print("Weekday")
````

## 9. Real-World Example: Student Grade System

````python
marks = 88

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
````

### Real-World Example: Student Grade System Using `match-case`

````python
marks = 88

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

match grade:
    case "A":
        print("Grade A - Excellent Performance")
    case "B":
        print("Grade B - Very Good Performance")
    case "C":
        print("Grade C - Good Performance")
    case "D":
        print("Grade D - Needs Improvement")
    case "F":
        print("Grade F - Failed")
    case _:
        print("Invalid Grade")
````

## Summary Table

| Statement | Purpose |
|-----------|---------|
| `if` | Execute code when a condition is True |
| `if-else` | Choose between two alternatives |
| `if-elif-else` | Choose among multiple alternatives |
| Nested `if` | Perform multi-level decision making |
| Logical Operators | Combine conditions |
| Ternary Operator | One-line conditional expression |
| `match-case` | Pattern matching / switch-case alternative |


## Key Points for Students

- Python uses **indentation** instead of braces `{}`.
- Conditions evaluate to **True** or **False**.
- `elif` is Python's equivalent of **else-if**.
- Use `and`, `or`, and `not` to combine conditions.
- `match-case` is available from **Python 3.10 onwards**.
- For a large number of choices, `match-case` is often cleaner than many `elif` statements.
- Always remember the colon (`:`) after `if`, `elif`, `else`, and `case` statements.
